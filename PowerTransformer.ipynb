```python
x_train, x_test, y_train, y_test = train_test_split(df.drop(columns= ['Strength']), df['Strength'], test_size = 0.2, random_state = 42)
x_train
```

```python
df.describe()
```

```python
#Applying Regression without any Transformation

lr = LinearRegression()
lr.fit(x_train,y_train)
y_pred = lr.predict(x_test)
r2_score(y_test, y_pred)
```

```python
np.mean(cross_val_score(lr,x_train,y_train , scoring = 'r2', cv = 20))
```

```python
#Applying Box-Cox transformation

pt = PowerTransformer(method = 'box-cox', standardize = True)
x_train_transformed = pt.fit_transform(x_train+0.000001)
x_test_transformed = pt.transform(x_test+0.000001)

pd.DataFrame({'cols': x_train.columns, 'box_cox_lambdas':pt.lambdas_})
```

```python
#Applying Regression after PowerTransformation
lr = LinearRegression()
lr.fit(x_train_transformed,y_train)

y_pred = lr.predict(x_test_transformed)
r2_score(y_test, y_pred )
```

```python
np.mean(cross_val_score(lr,x_train_transformed, y_train, scoring = 'r2', cv = 30))
```
